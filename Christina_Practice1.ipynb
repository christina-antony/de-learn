```python
%python
file_path = "/Volumes/workspace/ch_schema/volume_usage_metrics/employee.csv"

df1 = spark.read.option("header", "true").option("inferSchema", "true").csv(file_path)

display(df1)
```

```python
%python
# Write df1 as a managed Delta table (overwrites if the table already exists)
df1.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("default.employee")

# Verify the table was created and view the rows
display(spark.table("default.employee"))
```

```python
%python
display(spark.sql("SELECT * FROM default.employee LIMIT 2"))
```
